# Convolutional Neural Networks (CNN)

## Why ANN is Not Suitable for Images?

### 1. High Computational Cost
- Images must be flattened into a 1D vector.
- This creates a huge number of weights.
- Training becomes computationally expensive.

### 2. Overfitting
- Large number of parameters increases model complexity.
- The model may memorize training data instead of learning patterns.

### 3. Loss of Spatial Information
- Flattening destroys the relationship between neighboring pixels.
- Important image features may be lost.

---

# ANN Node vs CNN Filter

## ANN Node

Output = Activation(Σ(x × w) + b)

Where:
- x = Input
- w = Weight
- b = Bias

## CNN Filter

- A filter works on a small local region of the image.
- Performs dot product, adds bias, and applies activation.
- The same filter slides across the entire image.

### Key Difference
- ANN uses global connections.
- CNN uses local connections and weight sharing.

---

# Weight Sharing

- The same filter weights are used throughout the image.
- Reduces the number of parameters.
- Makes training faster.
- Helps reduce overfitting.

---

# Trainable Parameters Example

### Given:
- Input Image = 228 × 228 × 3
- Filters = 50
- Kernel Size = 3 × 3 × 3

### Calculation

Weights per Filter:

3 × 3 × 3 = 27

Total Weights:

27 × 50 = 1350

Biases:

50

Total Parameters:

1350 + 50 = 1400

**Total Trainable Parameters = 1400**

---

# Effect of Increasing Image Size

Even if image size changes from:

228 × 228 × 3

to

1080 × 1080 × 3

the number of trainable parameters remains **1400**.

Reason:
- Parameters depend on kernel size and number of filters.
- They do not depend on image dimensions.

---

# ANN vs CNN

| Feature | ANN | CNN |
|----------|----------|----------|
| Input | Flattened 1D Vector | 2D/3D Image |
| Connectivity | Fully Connected | Local Connectivity |
| Weights | Independent | Shared |
| Parameters | High | Low |
| Spatial Information | Lost | Preserved |
| Overfitting | More | Less |
| Best For | Tabular Data | Images |

---

# Conclusion

CNN is preferred for image processing because it preserves spatial information, uses weight sharing, requires fewer parameters, trains faster, and reduces overfitting.
