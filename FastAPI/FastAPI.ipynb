# 🚀 Why is FastAPI Fast?

FastAPI uses **ASGI**

Flask uses **WSGI**

## Flask (WSGI)

One request is processed at a time.

## FastAPI (ASGI)

Multiple requests can be processed simultaneously.

This makes FastAPI much faster.

# ASGI vs WSGI (Short Notes)

| WSGI | ASGI |
|------|------|
| **Web Server Gateway Interface** | **Asynchronous Server Gateway Interface** |
| Synchronous (Blocking) | Asynchronous (Non-Blocking) |
| Handles one request at a time | Handles multiple requests at the same time |
| Slower for many users | Faster and more scalable |
| Used by Flask | Used by FastAPI |
| Server: Gunicorn | Server: Uvicorn |
| Best for simple web apps | Best for APIs, ML deployment, and real-time apps |

### Easy Trick 🧠

**WSGI = Wait** → Waits for one request to finish.

**ASGI = Async** → Handles multiple requests concurrently.

# 📊 Flask vs FastAPI

| Flask | FastAPI |
|--------|----------|
| WSGI | ASGI |
| Synchronous | Asynchronous |
| Slower | Faster |
| Gunicorn | Uvicorn |
| More Code | Less Code |

# 📍 Path Parameters

Path Parameters are values written inside the URL.

Example

Here

is the Path Parameter.

They are used to access a specific resource.

````python
from fastapi import FastAPI

app = FastAPI()

@app.get("/patient/{patient_id}")
def get_patient(patient_id:str):
    return {"Patient ID":patient_id}
````

# 🛠 Path()

Path() is used to

- Add description
- Add examples
- Validate values

Example

patient_id = Path(

    example="P001"
)

# Understanding `Path()` in FastAPI

```python
patient_id = Path(
    ...,
    description="Patient ID",
    example="P001"
)
```

### 🔹 `patient_id`
- The **path parameter** (value taken from the URL).
- Example:
  ```
  /patient/P001
  ```
  Here, `patient_id = "P001"`.

---

### 🔹 `Path()`
- A FastAPI function used to **validate** and **document** path parameters.

---

### 🔹 `...` (Ellipsis)
- Means the parameter is **required**.
- The API will not work if this value is missing.

---

### 🔹 `description="Patient ID"`
- Adds a description for the parameter.
- This description appears in the **Swagger UI (`/docs`)**.

---

### 🔹 `example="P001"`
- Shows an example value in the API documentation.
- Helps users understand the expected input format.

---

## Example

```python
from fastapi import FastAPI, Path

app = FastAPI()

@app.get("/patient/{patient_id}")
def get_patient(
    patient_id: str = Path(
        ...,
        description="Unique ID of the patient",
        example="P001"
    )
):
    return {"Patient ID": patient_id}
```

### If the URL is:
```
/patient/P001
```

### Output:
```json
{
  "Patient ID": "P001"
}
```

# Query Parameters in FastAPI

## What are Query Parameters?

Query parameters are **optional values** passed in the URL **after the `?` symbol**.

They are used to:
- 🔍 Search
- 📂 Filter
- ↕️ Sort
- 📄 Pagination

---

## URL Syntax

```
/path?key1=value1&key2=value2
```

Example:

```
/sort?sort_by=bmi&order=ascending
```

Here:

- `sort_by = bmi`
- `order = ascending`

are **query parameters**.

---

## Example

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/sort")
def sort_data(
    sort_by: str,
    order: str = "ascending"
):
    return {
        "sort_by": sort_by,
        "order": order
    }
```

---

## Explanation of Terms

### `sort_by: str`
- A **required query parameter**.
- It tells the API which field to sort by.

Example:
```
sort_by=bmi
```

---

### `order: str = "ascending"`
- An **optional query parameter**.
- If the user doesn't provide it, the default value is `"ascending"`.

Example:
```
order=descending
```

If omitted:
```
order = "ascending"
```

---

## Example Requests

### URL 1
```
/sort?sort_by=bmi&order=descending
```

Output:
```json
{
    "sort_by": "bmi",
    "order": "descending"
}
```

---

### URL 2
```
/sort?sort_by=height
```

Output:
```json
{
    "sort_by": "height",
    "order": "ascending"
}
```

(Default value is used.)

---

## Using `Query()`

```python
from fastapi import Query

sort_by: str = Query(
    ...,
    description="Field to sort by",
    example="bmi"
)

order: str = Query(
    "ascending",
    description="Sorting order",
    example="descending"
)
```

### Explanation

- **`Query()`** → Used to validate and document query parameters.
- **`...` (Ellipsis)** → Makes the parameter **required**.
- **`"ascending"`** → Default value (optional parameter).
- **`description`** → Description shown in Swagger (`/docs`).
- **`example`** → Example value shown in Swagger.

---

## Quick Difference

| Path Parameter | Query Parameter |
|---------------|----------------|
| Inside URL path | After `?` |
| Usually required | Usually optional |
| Identifies a specific resource | Used for filtering, sorting, searching |
