# 🐍 Pydantic Notes (FastAPI)

## What is Pydantic?

**Pydantic** is a Python library used for **data validation** and **data parsing**.

It checks whether the input data is correct before using it in our application.

It is widely used with **FastAPI**.

---

## Why do we need Pydantic?

Python is a **dynamically typed language**.

Example:

```python
age = 20
age = "Twenty"
```

Python allows this.

But in APIs and databases, we need **fixed data types**.

Pydantic solves this problem.

---

## What Pydantic Does

✅ Checks data types

✅ Validates input

✅ Converts compatible data automatically

✅ Gives clear error messages

✅ Makes APIs more secure

Sometimes checking only the data type is **not enough**.

These are **business rules**.

Pydantic Validators help us implement these rules automatically.

# 📌 What is field_validator?

A **field_validator** validates or modifies **only one field**.

It is used when validation depends on a single field.

Examples:

- Validate email
- Validate age
- Convert name to uppercase
- Remove extra spaces

````python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):
    name:str
    @field_validator("name") #The validator will run only for the **name** field.
    @classmethod #Makes this function a class method
    def convert_upper(cls,value): # **cls** → Current class-> **value** → Value entered by the user
        return value.upper()



patient=Patient(
name="nitish")

print(patient)
````

````python
# 📌 Example 2: Validate Email Domain
#Suppose only company employees can register.
# Allowed domains:- hdfc.com- icici.com
#Any other domain should give an error.
from pydantic import BaseModel,field_validator

class Patient(BaseModel):
    email:str
    @field_validator("email")
    @classmethod
    def validate_email(cls,value):
        valid_domains=["hdfc.com","icici.com"]
        domain=value.split("@")[-1]
        if domain not in valid_domains:
            raise ValueError("Invalid Company Email")
        return value

patient=Patient(email="abc@hdfc.com")

print(patient)
````

# 📌 Multiple Field Validators

A model can have multiple validators.

Each validator works independently.

````python
from pydantic import BaseModel,field_validator

class Patient(BaseModel):
    name:str
    email:str
    @field_validator("name")
    @classmethod
    def upper_name(cls,value):
        return value.upper()

    @field_validator("email")
    @classmethod
    def validate_email(cls,value):
        if "@" not in value:
            raise ValueError("Invalid Email")
        return value
````

# 📌 What is model_validator?

field_validator validates **one field**.

model_validator validates **multiple fields together**.

It checks the relationship between two or more fields.

# 📌 When do we use model_validator?

Examples

✔ Age > 60

Emergency Contact Required

✔ Salary > 50000

PAN Card Required

✔ Student Age <18

Parent Contact Required

These conditions depend on **multiple fields**.

````python
from pydantic import BaseModel,model_validator
from typing import Dict

class Patient(BaseModel):
    name:str
    age:int
    contact_details:Dict[str,str]
    @model_validator(mode="after")
    def emergency_contact(self):
        if self.age>60 and "emergency" not in self.contact_details:
            raise ValueError("Emergency Contact Required")
        return self
````

# 📌 What does mode="after" mean?

Validation happens in two steps.

Step 1

Pydantic checks data types.

↓

Step 2

model_validator runs.

That's why we use

```python
mode="after"
```

# 📌 What are Computed Fields?

Sometimes a value is **not entered by the user**.

Instead, it is **calculated automatically** from other fields.

Example:

- BMI = Weight / Height²
- Total Price = Quantity × Price
- Age from Date of Birth

These are called **Computed Fields**.

````python
from pydantic import BaseModel, computed_field
class Patient(BaseModel):
    weight: float
    height: float
    @computed_field     # Creates a field automatically.
    @property     #Allows us to access bmi like a normal variable.
    def bmi(self) -> float:
        return round(self.weight / (self.height ** 2), 2)
````

# 📌 What is a Nested Model?

A Nested Model means

**One Pydantic model is used inside another model.**

Instead of storing everything in one class,

we divide data into smaller models.
# 📌 Why Use Nested Models?

✅ Better Organization

✅ Easy to Read

✅ Reusable

✅ Automatic Validation

Example

Instead of

Patient

- city

- pincode

Create

Address Model

and use it inside Patient.

````python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    state: str
    pincode: int


class Patient(BaseModel):
    name: str
    age: int
    address: Address
````
