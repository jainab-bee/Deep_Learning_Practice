```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

Padding adds extra rows and columns around the boundaries of the input matrix. Typically, these added borders are filled with zeros—a process known as Zero Padding

```python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

No information loss after add padding .... size of output matrix(feature map) from each convolutional matrix is 28*28 as input image size

```python
model.summary()
```


Stride refers to the number of pixels by which the filter shifts or jumps over the input matrix during convolution

```python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

after apply padding+strides .

Standard Feature Map Formula (With Padding & Stride)The spatial dimension of the resulting feature map is calculated as:$$\left\lfloor \frac{n + 2p - f}{s} \right\rfloor + 1$$

 $n$: Input Image dimension ($n \times n$)

$f$: Filter/Kernel dimension ($f \times f$)

$p$: Padding width

$s$: Stride value
